```python

def stats(data):

  # Check whether the arguement is a tuple. If it is, the function can continue
  # without any changes.
  if type(data) == tuple:
    pass

  # If the arguement is a set, convert it to a list so it can be sorted later
  # when finding the median and minimum values.
  elif type(data) == set:
    data = list(data)

  # If the arguement is not a tuple or a set, display a warning and stop the
  # function by returning none.
  else:
    print("Warning: argument of wrong type")
    return None

  # Check whether data contains any elements. If the length is 0, there isn't
  # any data to calculate the statistics from.
  if len(data) == 0:
      print("Warning: data is empty")
      return None

  # Create an empty dictionary that will be used to store the calculated
  # statistics before they are returned.
  output = {}

  # Calculate the mean by adding all of the values together and dividing by the
  # total number of values in data.
  output["mean"] = sum(data) / len(data)

  # Sort the values into ascending order. This makes it easy to find both the
  # minimum and median values.
  sorted_data = sorted(data)

  # Find the position of the middle element in the sorted data. Integer division
  # ensures the index is a whole number.
  middle = len(sorted_data) // 2

  # Save the middle value of the sorted data as the median.
  output["median"] = sorted_data[middle]

  # Save the first value in the sorted data as the minimum value.
  output["minimum"] = sorted_data[0]

  # Return the completed dictionary containing the mean, median and mimumum
  # values.
  return output

# Store the digits of the student ID in a tuple so that they can be passes into
# the stats function.
studentID = (1, 4, 4, 1, 3, 5, 5)

# Call the funtion using the student ID and save the returned dictionary in the
# variable stats_dict.
stats_dict = stats(studentID)

# Display the contents of the dictionary.
print(stats_dict)

```
